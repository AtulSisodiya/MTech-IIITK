```python
import networkx as nx
import matplotlib.pyplot as plt
import random
```

```python
G=nx.Graph()   #makes undirected graph
```

```python
cities = [
    'Mumbai', 'Delhi', 'Bangalore', 'Jaipur', 'Surat', 'Lucknow', 'Indore', 'Bhopal', 'Visakhapatnam', 'Patna'
]
```

```python
cost = [250, 500, 750, 1200, 1500, 2000, 3000, 3500, 4000, 5000]
```

```python
for each in cities: G.add_node(each)
```

```python
nx.draw(G, with_labels=1)
```

```python
G.nodes()
```

```python
l='Mumbai', 'Delhi', 'Bangalore', 'Jaipur', 'Surat', 'Lucknow', 'Indore', 'Bhopal', 'Visakhapatnam', 'Patna'
```

```python
c1=random.choice(l)
```

```python
# add edges randomly
while(G.number_of_edges()<10):
 startpoint=random.choice(l)
 endpoint=random.choice(l)
 if(startpoint!=endpoint) and G.has_edge(startpoint,endpoint)==0:
   w=random.choice(cost)
   G.add_edge(startpoint,endpoint,weight=w)
```

```python
pos=nx.circular_layout(G)
nx.draw(G,pos,with_labels=1)
```

```python
nx.draw_networkx_edge_labels(G,pos)
```

```python
nx.is_connected(G)
```

```python
for u in G.nodes():
 for v in G.nodes():
    print(u,v, nx.has_path(G,u,v))
```
